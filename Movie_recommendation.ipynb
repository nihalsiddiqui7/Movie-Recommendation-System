```python
import ast

def extract_names(text):
  try:
    for item in ast.literal_eval(text):
      return item['name'].replace(" ","")
  except:
    return[]
```

```python
def extract_cast_names(cast_list):
    try:
        return [person['name'].replace(" ", "") for person in cast_list[:3]]
    except:
        return []
```

```python
def convert_cast(text):
  l = []
  counter = 0
  for i in ast.literal_eval(text):
    if counter < 3:

      l.append(i['name'])
      counter +=1

  return l
```

```python
df['cast'] = df['cast'].apply(convert_cast)
```

```python
df.head()
```

```python
def convert(text):
  l=[]
  for i in ast.literal_eval(text):
    l.append(i['name'])

  return l
```

```python
df['keywords'] = df['keywords'].apply(convert)
```

```python
df.head()
```

```python
def fetch_director(text):
  l=[]
  for i in ast.literal_eval(text):
    if i['job'] == 'Director':
      l.append(i['name'])
      break

  return l
```

```python
df['director'] = df['crew'].apply(fetch_director)
```

```python
df.head()
```

```python
df['overview']= df['overview'].apply(lambda x :x.split())
```

```python
def remove_spaces(word):
  l=[]
  for i in word:
    l.append(i.replace(" ",""))

  return l
```

```python
df['director'] = df['director'].apply(remove_spaces)
df['keywords'] = df['keywords'].apply(remove_spaces)
df['cast'] = df['cast'].apply(remove_spaces)


```

```python
# If overview is a string: turn it into a list of words
df['overview'] = df['overview'].fillna('').apply(lambda x: x.split() if isinstance(x, str) else x)
```

```python
df.head()
```

```python
# If genres is a string: turn it into a list of words
df['genres'] = df['genres'].fillna('').apply(lambda x: x.split() if isinstance(x, str) else x)
```

```python
df["tags"] = df["overview"] + df["director"]+ df["cast"] + df["genres"] + df["keywords"]
```

```python
df.iloc[0]['tags']
```

```python
final_df = df[["tags","title","movie_id"]]
```

```python
final_df.head()
```

```python
def flatten_and_join(x):
    # Only keep items that are strings
    flat_list = [str(item) for item in x if isinstance(item, str)]
    return " ".join(flat_list)

final_df['tags'] = final_df['tags'].apply(flatten_and_join)
```

```python
final_df.head()
```

```python
final_df["tags"]= final_df['tags'].apply(lambda x:x.lower())
```

```python
final_df.shape
```

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
def stems(text):
  l=[]
  for i in text.split():
    l.append(stemmer.stem(i))
  return " ".join(l)
```

```python
final_df["tags"]= final_df['tags'].apply(stems)
```

```python
final_df.iloc[0]['tags']
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000,stop_words="english")
```

```python
vector = cv.fit_transform(final_df['tags']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
def recommender(movie):
  index = final_df[final_df['title']== movie].index[0]
  distances = sorted(list(enumerate(similarity[index])),reverse=True ,key =lambda x: x[1])
  print("Recommended Movies:\n")

  for i in distances[1:11]:
    print(final_df.iloc[i[0]].title)
```

```python
recommender("Spectre")
```
